```python
# Load all six datasets

from pathlib import Path
import pandas as pd

# Locate the project and data folders
PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / "data" / "raw"

# Handle CSV encodings
def read_csv_safely(filepath):
    try:
        return pd.read_csv(filepath, encoding="utf-8-sig")
    except UnicodeDecodeError:
        return pd.read_csv(filepath, encoding="cp1252")

# Load all six datasets
customers = read_csv_safely(DATA_DIR / "Customers.csv")
sales = read_csv_safely(DATA_DIR / "Sales.csv")
products = read_csv_safely(DATA_DIR / "Products.csv")
stores = read_csv_safely(DATA_DIR / "Stores.csv")
exchange_rates = read_csv_safely(DATA_DIR / "Exchange_Rates.csv")
dictionary = read_csv_safely(DATA_DIR / "Data_Dictionary.csv")

print("All six datasets loaded successfully.")
```

Another formatting issue!
The value "1,060.22" contains a comma as a thousands separator. The dollar sign was removed, but not the comma, so pd.to_numeric() couldn't interpret it as a number.

```python
# Clean and convert Unit Cost USD
products["Unit Cost USD"] = pd.to_numeric(
    products["Unit Cost USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# Clean and convert Unit Price USD
products["Unit Price USD"] = pd.to_numeric(
    products["Unit Price USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# Verify the conversion
print(products[["Unit Cost USD", "Unit Price USD"]].dtypes)

print("\nFirst 10 converted values:")
print(products[["Unit Cost USD", "Unit Price USD"]].head(10))
```

### State Code Import Issue

During the initial inspection, 10 values in the `State Code` column appeared to be missing. Upon further investigation of the original CSV file, it was observed that these values were actually the literal state code `NA`, associated with customers in Napoli, Italy.

Pandas had interpreted `NA` as a missing value during the initial import because it is included among its default missing-value indicators.

The Customers dataset was reloaded with `keep_default_na=False` and `na_values=[""]` to preserve literal `NA` codes while still treating empty fields as missing. The corrected dataset contains the original state codes without imputing or inventing values.

Investigate missing delivery dates

```python
print("Total sales rows:", len(sales))

print(
    "\nMissing Delivery Date by Order Date year:"
)
print(
    sales.assign(
        OrderYear=sales["Order Date"].dt.year,
        DeliveryMissing=sales["Delivery Date"].isna()
    )
    .groupby("OrderYear")["DeliveryMissing"]
    .agg(["sum", "count"])
    .rename(columns={
        "sum": "Missing Delivery Dates",
        "count": "Total Sales Rows"
    })
)
```

```python
# Calculate missing delivery dates by order year
delivery_missing_by_year = (
    sales.assign(
        OrderYear=sales["Order Date"].dt.year,
        DeliveryMissing=sales["Delivery Date"].isna()
    )
    .groupby("OrderYear")
    .agg(
        Total_Sales_Rows=("DeliveryMissing", "size"),
        Missing_Delivery_Dates=("DeliveryMissing", "sum")
    )
)

delivery_missing_by_year["Missing_Percentage"] = (
    delivery_missing_by_year["Missing_Delivery_Dates"]
    / delivery_missing_by_year["Total_Sales_Rows"]
    * 100
).round(2)

print(delivery_missing_by_year)
```

```python
# Summarize delivery-date completeness at the order level
order_delivery = (
    sales.groupby("Order Number")
    .agg(
        Total_Line_Items=("Line Item", "size"),
        Missing_Delivery_Lines=("Delivery Date", lambda x: x.isna().sum()),
        Recorded_Delivery_Lines=("Delivery Date", lambda x: x.notna().sum())
    )
)

order_delivery["Delivery_Date_Status"] = "All dates recorded"

order_delivery.loc[
    order_delivery["Missing_Delivery_Lines"].eq(
        order_delivery["Total_Line_Items"]
    ),
    "Delivery_Date_Status"
] = "All dates missing"

order_delivery.loc[
    order_delivery["Missing_Delivery_Lines"].gt(0)
    & order_delivery["Recorded_Delivery_Lines"].gt(0),
    "Delivery_Date_Status"
] = "Mixed"

print(order_delivery["Delivery_Date_Status"].value_counts())

print("\nPercentage of orders in each category:")
print(
    (order_delivery["Delivery_Date_Status"].value_counts(normalize=True) * 100)
    .round(2)
)
```

### Missing Delivery Dates

The `Delivery Date` column contains 49,719 missing values across 62,884 sales line items. Missing delivery dates occur in every order year examined.

An order-level investigation identified 26,326 distinct orders: 20,746 orders had no delivery date recorded for any line item, while 5,580 orders had delivery dates recorded for all line items. No orders contained a mixture of missing and recorded delivery dates.

Missing delivery dates were retained as missing because the available data does not establish why they are absent. They will not be imputed. Delivery-time analysis will use records with valid order and delivery dates, and the limitation will be disclosed when presenting results.
