Inspect the dataset to understand the structure, size, and quality of each file before cleaning.
All six files will be inspected in one go rather than opening them individually. This saves time and gives a clear overview of the dataset.

```python
import pandas as pd

# Store all six CSV files in a dictionary
data_files = {
    "Customers": DATA_DIR / "Customers.csv",
    "Data_Dictionary": DATA_DIR / "Data_Dictionary.csv",
    "Exchange_Rates": DATA_DIR / "Exchange_Rates.csv",
    "Products": DATA_DIR / "Products.csv",
    "Sales": DATA_DIR / "Sales.csv",
    "Stores": DATA_DIR / "Stores.csv"
}

# Inspect each file
for name, filepath in data_files.items():
    df = pd.read_csv(filepath)

    print(f"FILE: {name}")
    print("_" * 60)

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 3 rows:")
    print(df.head(3))

    print("\n")
```

Why did an error occur?
        UnicodeDecodeError: 'utf-8' codec can't decode byte 0xfc

Python is trying to read the CSV files using UTF-8 encoding, but the file contains a character that isn't valid under that encoding.
The byte 0xfc commonly represents the character ü in Latin-1 encoding. 

[A version of the Global Electronics Retailer dataset has been documented as having a Customers.csv file encoded in Latin-1. (GitHub)]

This encoding issue can be fixed by telling pandas which encoding to use.

First, create a function `read_csv_safely()` to handle the encoding differences.
Use utf-8-sig to read UTF-8 files and handle any optional byte-order mark. cp1252 will be the fallback when UTF-8 decoding fails and the rest of the code inspects each dataset as before.

```python
import pandas as pd

# Read CSV files while handling common text encodings
def read_csv_safely(filepath):
    try:
        return pd.read_csv(filepath, encoding="utf-8-sig")
    except UnicodeDecodeError:
        return pd.read_csv(filepath, encoding="cp1252")


# Inspect all six files
for name, filepath in data_files.items():

    df = read_csv_safely(filepath)

    print(f"FILE: {name}")
    print("_" * 60)

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 3 rows:")
    print(df.head(3))

    print()
```

```python
# Check missing values and duplicates

for name, filepath in data_files.items():

    df = read_csv_safely(filepath)

    print(f"DATA QUALITY: {name}")
    print("_" * 50)

    # Count missing values
    missing = df.isnull().sum()

    print("\nMissing values:")
    if missing.sum() == 0:
        print("No missing values found.")
    else:
        print(missing[missing > 0])

    # Count completely duplicated rows
    print("\nDuplicate rows:", df.duplicated().sum())

    print()
```
